## How to Obtain GITHUB_TOKEN and Verify with `chatlas`

This notebook aims as instructions on how to obtain your own `GITHUB_TOKEN` for use with `querychat` and `chatlas` in this repository.

### Step 1:

Click your profile picture on the top right corner of your personal GitHub and a menu bar should appear.

### Step 2:

Click on `Settings` from the menu bar. This should proceed to the `Settings` page.

### Step 3:

Scroll down on the `Settings` page and select `Developer Settings`. This should lead you to the `Developer Page`

### Step 4:

Click `Personal access token` on the side bar on the left which should expand. Select `Fine-grained tokens` within this tab.

### Step 5:

Once on the `Fine-grained personal access tokens` page, select `Generate new token` on the right top corner.

### Step 6:

This should lead you to generate a new token. Once the token is revealed.

**NOTE:** You must copy the token once revealed. This is the only time you get to access this token. If you navigate to another page and did not copy the token, you must create a new one.

Refer to the demonstration:

![gh_token_demo](../img/gh_token.gif)

### Step 7:

Create a new file **at the root** of the repo called `.env`. 

Within it, paste your token in double quotes to the `GITHUB_TOKEN` variable.

Your `.env` should have the following contents:

```python
GITHUB_TOKEN="<paste_your_token_here>"
```

### Step 8:

Verify your API key by running the following cell (use the `fin-health` environment for this notebook):

````python
from chatlas import ChatGithub
from dotenv import load_dotenv

load_dotenv()

chat = ChatGithub(model="gpt-4.1-mini")
chat.chat("Hello")
````

If done correctly, you should have a response from the LLM. If not, you will either get no response from the LLM, or the code will raise an error that it cannot find your API key.
